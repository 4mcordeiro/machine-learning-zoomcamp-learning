```python
carros_asia = df[df['origin'] == 'Asia']
carros_asia
```

```python
carros_asia = carros_asia[['vehicle_weight', 'model_year']]
carros_asia
```

```python
carros_asia = carros_asia.head(7)
carros_asia
```

```python
X = carros_asia.values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```
