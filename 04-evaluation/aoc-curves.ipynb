## AUC para avaliar atributos 📊

Percebi que dá para usar a AUC para ver o quanto um atributo, sozinho, consegue separar as classes. É rápido e fácil.

A ideia: use a própria variável como se fosse a "previsão" e compare com o resultado real.

```python
import numpy as np
from sklearn.metrics import roc_auc_score
```

```python
# Dados do exemplo
glicemia = np.array([85, 90, 95, 105, 110, 115, 125, 130, 140, 150])
diagnostico = np.array([0, 0, 0, 0, 1, 0, 1, 1, 1, 1])

# AUC univariada (isolada)
auc_univariada = roc_auc_score(diagnostico, glicemia)

print(f"A AUC Univariada da Glicemia é: {auc_univariada:.2f}")
# A AUC Univariada da Glicemia é: 0.96
```

**Como ler o resultado:**
- 0,5 → o atributo não separa nada (equivale a um chute)
- perto de 1 → separa muito bem
- abaixo de 0,5 → a relação é inversa; use `-variavel` para inverter

Quanto maior a AUC, mais o atributo ajuda a distinguir as classes.
