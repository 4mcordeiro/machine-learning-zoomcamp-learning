## AUC para avaliar atributos 📊

Na 4º aula do #machine-learning-zoomcamp da DataTalks percebi que dá para usar a AUC para ver o quanto um atributo, sozinho, consegue separar as classes. É rápido e fácil.

A ideia é usar própria variável como se fosse a "previsão" e comparar com o resultado real.

```python
import numpy as np
from sklearn.metrics import roc_auc_score
```

```python
# Dados do exemplo
horas_estudo = np.array([1, 2, 3, 4, 5, 6, 7, 8])
aprovado = np.array([0, 0, 0, 1, 0, 1, 1, 1])

# AUC univariada (isolada)
auc_univariada = roc_auc_score(aprovado, horas_estudo)

print(f"A AUC Univariada das Horas de Estudo é: {auc_univariada:.2f}")
```

**Como ler o resultado:**
- 0,5 → o atributo não separa nada (equivale a um chute)
- perto de 1 → separa muito bem
- abaixo de 0,5 → a relação é inversa; use `-variavel` para inverter

Quanto maior a AUC, mais o atributo ajuda a distinguir as classes.
